```python
import pandas as pd
```

```python
columns = [
    "filename",
    "satellite_code",
    "satellite_name",
    "status",
    "confidence",
    "metadata_location",
    "metadata_orientation",
    "crop_name",
    "read_string",
    "classifier_method",
    "details",
    "processing_seconds"
]

# Load CSV, discarding the first 4 rows
df = pd.read_csv(
    r"..\output\alouette_2_test_predictions.csv",
    skiprows=4,
    header=None,
    names=columns
)

# Clean possible extra spaces
df["satellite_name"] = df["satellite_name"].astype(str).str.strip()

# Ground truth label
true_label = "Alouette 2"

# Calculate accuracy
accuracy = (df["satellite_name"] == true_label).mean()

print(f"Accuracy: {accuracy:.4f}")
print(f"Correct: {(df['satellite_name'] == true_label).sum()} / {len(df)}")
```
